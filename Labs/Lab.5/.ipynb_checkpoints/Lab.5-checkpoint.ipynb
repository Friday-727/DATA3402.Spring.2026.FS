1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
class matrix:
    def __init__(self, n=None, m=None, data=None):

        if data is not None:
            if not all(isinstance(row, list) for row in data):
                raise TypeError("data must be a 2D list")

            if len(data) == 0:
                raise ValueError("data cannot be empty")

            col_len = len(data)
            row_len = len(data[0])

            for row in data:
                if len(row) != row_len:
                    raise ValueError("All rows must have the same amount of elements")

            self.n_rows = col_len
            self.n_cols = row_len
            self.data = data

        elif isinstance(n, int) and isinstance(m, int):
                self.n_rows = n
                self.n_cols = m
                self.data = [[0 for _ in range(m)] for _ in range(n)]

        else:
            raise ValueError("Provide either data OR dimensions (n, m)")

    #a function that allows for slicing the class
    def __getitem__(self, index):
    
            if isinstance(index, tuple):
                if len(index) != 2:
                    raise IndexError("Use A[i, j] for indexing")
    
                i, j = index
                return self.data[i][j]
    
            elif isinstance(index, int):
                return self.data[index]
    
            else:
                raise TypeError("Invalid index type")

    # a method to allow one matrix of a particular size to be assigned to another matrix of the same size
    def assign(self, other):
        if not isinstance(other, matrix):
            raise TypeError("You can only assign the new matrix from another matrix")

        if self.n_rows != other.n_rows or self.n_cols != other.n_cols:
            raise ValueError("Matrices must be of the same dimensions")

        self.data = [row[:] for row in other.data]
```

2. Add the following methods:
    * `shape()`: returns a tuple `(n,m)` of the shape of the matrix.
    * `transpose()`: returns a new matrix instance which is the transpose of the matrix.
    * `row(n)` and `column(n)`: that return the nth row or column of the matrix M as a new appropriately shaped matrix object.
    * `to_list()`: which returns the matrix as a list of lists.
    *  `block(n_0,n_1,m_0,m_1)` that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows. 
    * Modify `__getitem__` implemented above to support slicing.
        

```python
class matrix:
    def __init__(self, n=None, m=None, data=None):

        if data is not None:
            if not all(isinstance(row, list) for row in data):
                raise TypeError("data must be a 2D list")

            if len(data) == 0:
                raise ValueError("data cannot be empty")

            col_len = len(data)
            row_len = len(data[0])
            
            for row in data:
                if len(row) != row_len:
                    raise ValueError("All rows must have the same amount of elements")

            self.n_rows = col_len
            self.n_cols = row_len
            self.data = data

        elif isinstance(n, int) and isinstance(m, int):
            self.n_rows = n
            self.n_cols = m
            self.data = [[0 for _ in range(m)] for _ in range(n)]

        else:
            raise ValueError("Provide either data OR dimensions (n, m)")

    #a function that allows for slicing the class (__getitem__ supports slicing here)
    def __getitem__(self, index):
    
            if isinstance(index, tuple):
                if len(index) != 2:
                    raise IndexError("Use A[i, j] for indexing")
    
                i, j = index
                return self.data[i][j]
    
            elif isinstance(index, int):
                return self.data[index]
    
            else:
                raise TypeError("Invalid index type")

    # a method to allow one matrix of a particular size to be assigned to another matrix of the same size
    def assign(self, other):
        if not isinstance(other, matrix):
            raise TypeError("You can only assign the new matrix from another matrix")

        if self.n_rows != other.n_rows or self.n_cols != other.n_cols:
            raise ValueError("Matrices must be of the same dimensions")

        self.data = [row[:] for row in other.data]

    # a method that returns a tuple of (n,m)
    def shape(self):
        n = self.n_rows
        m = self.n_cols
        size_list = [n]
        size_list.append(m)
        size_tup = tuple(size_list)
        return(size_tup)

    # a method that returns a new matrix that is the transpose of the original one
    def transpose(self):
        new_matrix = [[self.data[i][j] for i in range(self.n_rows)] for j in range(self.n_cols)]
        return matrix(data = new_matrix)

    # method that returns the nth row or column of the matrix M as a new appropriately shaped matrix object
    def row(self, n):
        self.n = n
        index_matrix = self.data[n]
        return matrix(data=[index_matrix])
    
    def col(self, m):
        self.m = m
        index_matrix = [row[m] for row in self.data]
        return matrix(data=[index_matrix])

    # a method which returns the matrix as a list of lists
    def to_list(self):
        matrix_as_list = [[self.data[i][j] for i in range(self.n_rows)] for j in range(self.n_cols)]
        return matrix_as_list

    # a method that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows
    def block(self, n_0, n_1, m_0, m_1):
        self.n_0 = n_0
        self.n_1 = n_1
        self.m_0 = m_0
        self.m_1 = m_1

        if n_0 != m_0:
            raise ValueError("You must choose an appropriate range to dissect, for example: matrix A --> 2 to 4 matrix B --> 2 to 4")
        if n_1 != m_1:
            raise ValueError("You must choose an appropriate range to dissect, for example: matrix A --> 2 to 4 matrix B --> 2 to 4")

        else:
            submatrix = [row[m_0 : m_1] for row in self.data[n_0 : n_1]]
            
            return matrix(data=submatrix)
```

```python
def constant(n,m,c):

    if not isinstance(c, int):
        raise TypeError("the input value must be an integer")

    data = []
    for num in range(n):
        row = [c]*m
        data.append(row)
    return matrix(data = data)
```

4. Add the following member functions to your class. Make sure to appropriately test the dimensions of the matrices to make sure the operations are correct.
    * `M.scalarmul(c)`: a matrix that is scalar product $cM$, where every element of $M$ is multiplied by $c$.
    * `M.add(N)`: adds two matrices $M$ and $N$. Don’t forget to test that the sizes of the matrices are compatible for this and all other operations.
    * `M.sub(N)`: subtracts two matrices $M$ and $N$.
    * `M.mat_mult(N)`: returns a matrix that is the matrix product of two matrices $M$ and $N$.
    * `M.element_mult(N)`: returns a matrix that is the element-wise product of two matrices $M$ and $N$.
    * `M.equals(N)`: returns true/false if $M==N$.

```python
class matrix:
    def __init__(self, n=None, m=None, data=None):

        if data is not None:
            if not all(isinstance(row, list) for row in data):
                raise TypeError("data must be a 2D list")

            if len(data) == 0:
                raise ValueError("data cannot be empty")

            col_len = len(data)
            row_len = len(data[0])
            
            for row in data:
                if len(row) != row_len:
                    raise ValueError("All rows must have the same amount of elements")

            self.n_rows = col_len
            self.n_cols = row_len
            self.data = data

        elif isinstance(n, int) and isinstance(m, int):
            self.n_rows = n
            self.n_cols = m
            self.data = [[0 for _ in range(m)] for _ in range(n)]

        else:
            raise ValueError("Provide either data OR dimensions (n, m)")

    #a function that allows for slicing the class (__getitem__ supports slicing here)
    def __getitem__(self, index):
    
            if isinstance(index, tuple):
                if len(index) != 2:
                    raise IndexError("Use A[i, j] for indexing")
    
                i, j = index
                return self.data[i][j]
    
            elif isinstance(index, int):
                return self.data[index]
    
            else:
                raise TypeError("Invalid index type")

    # a method to allow one matrix of a particular size to be assigned to another matrix of the same size
    def assign(self, other):
        if not isinstance(other, matrix):
            raise TypeError("You can only assign the new matrix from another matrix")

        if self.n_rows != other.n_rows or self.n_cols != other.n_cols:
            raise ValueError("Matrices must be of the same dimensions")

        self.data = [row[:] for row in other.data]

    # a method that returns a tuple of (n,m)
    def shape(self):
        n = self.n_rows
        m = self.n_cols
        size_list = [n]
        size_list.append(m)
        size_tup = tuple(size_list)
        return(size_tup)

    # a method that returns a new matrix that is the transpose of the original one
    def transpose(self):
        new_matrix = [[self.data[i][j] for i in range(self.n_rows)] for j in range(self.n_cols)]
        return matrix(data = new_matrix)

    # method that returns the nth row or column of the matrix M as a new appropriately shaped matrix object
    def row(self, n):
        self.n = n
        index_matrix = self.data[n]
        return matrix(data=[index_matrix])
    
    def col(self, m):
        self.m = m
        index_matrix = [row[m] for row in self.data]
        return matrix(data=[index_matrix])

    # a method which returns the matrix as a list of lists
    def to_list(self):
        matrix_as_list = [[self.data[i][j] for i in range(self.n_rows)] for j in range(self.n_cols)]
        return matrix_as_list

    # a method that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows
    def block(self, n_0, n_1, m_0, m_1):
        self.n_0 = n_0
        self.n_1 = n_1
        self.m_0 = m_0
        self.m_1 = m_1

        if n_0 != m_0:
            raise ValueError("You must choose an appropriate range to dissect, for example: matrix A --> 2 to 4 matrix B --> 2 to 4")
        if n_1 != m_1:
            raise ValueError("You must choose an appropriate range to dissect, for example: matrix A --> 2 to 4 matrix B --> 2 to 4")

        else:
            submatrix = [row[m_0 : m_1] for row in self.data[n_0 : n_1]]
            
            return matrix(data=submatrix)



    # The next section entails matrix mathematical operations

    def scalarmul(self, c):
        self.c = c  #constant supplied by user
        result = []
        
        for row in self.data:
            new_row = []
            for num in row:
                new_row.append(num*c)
            result.append(new_row)
        
        return matrix(data=result)

    def add(self, other):
        result = []
        if self.shape() != other.shape():
            raise ValueError("To do matrix addition, both matrices must be of the same dimension")
        
        for i in range(len(self.data)):
            new_row = []

            for j in range(len(self.data[0])):
                new_row.append(self.data[i][j] + other.data[i][j])
               
            result.append(new_row)

        return matrix(data=result)

    def sub(self, other):
        result = []
        if self.shape() != other.shape():
            raise ValueError("To do matrix subtraction, both matrices must be of the same dimension")
        
        for i in range(len(self.data)):
            new_row = []

            for j in range(len(self.data[0])):
                new_row.append(self.data[i][j] - other.data[i][j])
               
            result.append(new_row)

        return matrix(data=result)

    def element_mult(self, other):
        result = []
        if self.shape() != other.shape():
            raise ValueError("To do matrix multiplication, both matrices must be of the same dimension")
        
        for i in range(len(self.data)):
            new_row = []

            for j in range(len(self.data[0])):
                new_row.append(self.data[i][j] * other.data[i][j])
               
            result.append(new_row)

        return matrix(data=result)

    
    def matmul(self, other):

    # Must multiply by another Matrix
        if not isinstance(other, matrix):
            raise TypeError("Can only multiply by another Matrix")
    
        # Dimension check #1
        if self.n_cols != other.n_rows or self.n_rows != other.n_cols:
            raise ValueError("Number of columns of first matrix must equal rows of second")
    
        # Create result matrix
        result = matrix(self.n_rows, other.n_cols)
        
        # Perform multiplication
        for i in range(self.n_rows):
            for j in range(other.n_cols):
                total = 0
                for k in range(self.n_cols):
                     total += self.data[i][k] * other.data[k][j]
                result.data[i][j] = total
        
        return result

    def eq(self, other):
        if self.data != other.data:
            return False
        else:
            return True
    
```

```python
class matrix:
    def __init__(self, n=None, m=None, data=None):

        if data is not None:
            if not all(isinstance(row, list) for row in data):
                raise TypeError("data must be a 2D list")

            if len(data) == 0:
                raise ValueError("data cannot be empty")

            col_len = len(data)
            row_len = len(data[0])
            
            for row in data:
                if len(row) != row_len:
                    raise ValueError("All rows must have the same amount of elements")

            self.n_rows = col_len
            self.n_cols = row_len
            self.data = data

        elif isinstance(n, int) and isinstance(m, int):
            self.n_rows = n
            self.n_cols = m
            self.data = [[0 for _ in range(m)] for _ in range(n)]

        else:
            raise ValueError("Provide either data OR dimensions (n, m)")

    #a function that allows for slicing the class (__getitem__ supports slicing here)
    def __getitem__(self, index):
    
            if isinstance(index, tuple):
                if len(index) != 2:
                    raise IndexError("Use A[i, j] for indexing")
    
                i, j = index
                return self.data[i][j]
    
            elif isinstance(index, int):
                return self.data[index]
    
            else:
                raise TypeError("Invalid index type")

    # a method to allow one matrix of a particular size to be assigned to another matrix of the same size
    def assign(self, other):
        if not isinstance(other, matrix):
            raise TypeError("You can only assign the new matrix from another matrix")

        if self.n_rows != other.n_rows or self.n_cols != other.n_cols:
            raise ValueError("Matrices must be of the same dimensions")

        self.data = [row[:] for row in other.data]

    # a method that returns a tuple of (n,m)
    def shape(self):
        n = self.n_rows
        m = self.n_cols
        size_list = [n]
        size_list.append(m)
        size_tup = tuple(size_list)
        return(size_tup)

    # a method that returns a new matrix that is the transpose of the original one
    def transpose(self):
        new_matrix = [[self.data[i][j] for i in range(self.n_rows)] for j in range(self.n_cols)]
        return matrix(data = new_matrix)

    # method that returns the nth row or column of the matrix M as a new appropriately shaped matrix object
    def row(self, n):
        self.n = n
        index_matrix = self.data[n]
        return matrix(data=[index_matrix])
    
    def col(self, m):
        self.m = m
        index_matrix = [row[m] for row in self.data]
        return matrix(data=[index_matrix])

    # a method which returns the matrix as a list of lists
    def to_list(self):
        matrix_as_list = [[self.data[i][j] for i in range(self.n_rows)] for j in range(self.n_cols)]
        return matrix_as_list

    # a method that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows
    def block(self, n_0, n_1, m_0, m_1):
        self.n_0 = n_0
        self.n_1 = n_1
        self.m_0 = m_0
        self.m_1 = m_1

        if n_0 != m_0:
            raise ValueError("You must choose an appropriate range to dissect, for example: matrix A --> 2 to 4 matrix B --> 2 to 4")
        if n_1 != m_1:
            raise ValueError("You must choose an appropriate range to dissect, for example: matrix A --> 2 to 4 matrix B --> 2 to 4")

        else:
            submatrix = [row[m_0 : m_1] for row in self.data[n_0 : n_1]]
            
            return matrix(data=submatrix)



    # The next section entails matrix mathematical operations

    def scalarmul(self, c):
        self.c = c  #constant supplied by user
        result = []
        
        for row in self.data:
            new_row = []
            for num in row:
                new_row.append(num*c)
            result.append(new_row)
        
        return matrix(data=result)

    def add(self, other):
        result = []
        if self.shape() != other.shape():
            raise ValueError("To do matrix addition, both matrices must be of the same dimension")
        
        for i in range(len(self.data)):
            new_row = []

            for j in range(len(self.data[0])):
                new_row.append(self.data[i][j] + other.data[i][j])
               
            result.append(new_row)

        return matrix(data=result)

    def sub(self, other):
        result = []
        if self.shape() != other.shape():
            raise ValueError("To do matrix subtraction, both matrices must be of the same dimension")
        
        for i in range(len(self.data)):
            new_row = []

            for j in range(len(self.data[0])):
                new_row.append(self.data[i][j] - other.data[i][j])
               
            result.append(new_row)

        return matrix(data=result)

    def element_mult(self, other):
        result = []
        if self.shape() != other.shape():
            raise ValueError("To do elemental matrix multiplication, both matrices must be of the same dimension")
        
        for i in range(len(self.data)):
            new_row = []

            for j in range(len(self.data[0])):
                new_row.append(self.data[i][j] * other.data[i][j])
               
            result.append(new_row)

        return matrix(data=result)


    #def mat_mult(self, other):
    def eq(self, other):
        if self.data != other.data:
            return False
        else:
            return True
        

    # Below are python operater overloading methods

    def __add__(self,other):
        result = []
        if self.shape() != other.shape():
            raise ValueError("To do matrix addition, both matrices must be of the same dimension")
        
        for i in range(len(self.data)):
            new_row = []

            for j in range(len(self.data[0])):
                new_row.append(self.data[i][j] + other.data[i][j])
               
            result.append(new_row)

        return matrix(data=result)

    def __sub__(self,other):
        result = []
        if self.shape() != other.shape():
            raise ValueError("To do matrix subtraction, both matrices must be of the same dimension")
        
        for i in range(len(self.data)):
            new_row = []

            for j in range(len(self.data[0])):
                new_row.append(self.data[i][j] - other.data[i][j])
               
            result.append(new_row)

        return matrix(data=result)
    
    def __eq__(self, other):
        if self.data != other.data:
            return False
        else:
            return True

    
    def __rmul__(self, other):
        self.c = other  #constant supplied by user
        result = []
        
        for row in self.data:
            new_row = []
            for num in row:
                new_row.append(num*other)
            result.append(new_row)
        
        return matrix(data=result)

    #overload matmult
    
    def __assign__(self, other):
        if not isinstance(other, matrix):
            raise TypeError("You can only assign the new matrix from another matrix")

        if self.n_rows != other.n_rows or self.n_cols != other.n_cols:
            raise ValueError("Matrices must be of the same dimensions")

        self.data = [row[:] for row in other.data]

    # matrix constant and matrix to matrix multiplication initialized by same overloading parameter

    def __mul__(self, other):
        
        if type(other) == int:
            self.c = other  #constant supplied by user
            result = []
            
            for row in self.data:
                new_row = []
                for num in row:
                    new_row.append(num*other)
                result.append(new_row)
            
            return matrix(data=result)

        else:
            if not isinstance(other, matrix):
                raise TypeError("Can only multiply by another Matrix")
        
            # Dimension check #1
            if self.n_cols != other.n_rows or self.n_rows != other.n_cols:
                raise ValueError("Number of columns of first matrix must equal rows of second")
        
            # Create result matrix
            result = matrix(self.n_rows, other.n_cols)
            
            # Perform multiplication
            for i in range(self.n_rows):
                for j in range(other.n_cols):
                    total = 0
                    for k in range(self.n_cols):
                         total += self.data[i][k] * other.data[k][j]
                    result.data[i][j] = total
            
            return result
```
