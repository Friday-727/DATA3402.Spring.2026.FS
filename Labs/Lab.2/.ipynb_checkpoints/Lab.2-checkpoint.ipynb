````python
# making the initial board
def make_board_(size):
    board = list()
    for i in range(size):
        row = list()
        for j in range(size):
            row.append(empty)   # for j in range size, the list holding the rows adds an "empty" value
        board.append(row)   # for i in range size, the list holding the board adds the rows declared in the line prior
    return board
````

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here

def who_won(board):
    
    def check_rows(board):
        for i in board:
            if all(j == player_1 for j in i):
                return 1
            elif all(j == player_2 for j in i):
                return 2

    def check_columns(board):
    # transposing the board by creating a list of lists
        transpose_board = [[row[k] for row in board] for k in range(len(board[0]))]
        for i in transpose_board:
            if all(j == player_1 for j in i):
                return 1
            elif all(j == player_2 for j in i):
                return 2
    
    def check_diagonal_1(board):
        diagonal_elements = list()
        for i in range(len(board)):
            diagonal_elements.append(board[i][i])
    
        if all(j == player_1 for j in diagonal_elements):
            return 1
        elif all(j == player_2 for j in diagonal_elements):
            return 2
    
    def check_diagonal_2(board):
        diagonal_elements = list()
        for i in range(len(board)):
            j = len(board) - 1
            diagonal_elements.append(board[i][j-i])
        for k in board:
            if all(j == player_1 for j in k):
                return 1
            elif all(j == player_2 for j in k):
                return 2

    if check_rows(board) == 1:
        return 1
    elif check_columns(board) == 1:
        return 1
    elif check_diagonal_1(board) == 1:
        return 1
    elif check_diagonal_2(board) == 1:
        return 1
    elif check_rows(board) == 2:
        return 2
    elif check_columns(board) == 2:
        return 2
    elif check_diagonal_1(board) == 2:
        return 2
    elif check_diagonal_2(board) == 2:
        return 2
    else:
        return -1
````

````python
n = 5  #In the game code, this will be explicit within the function as opposed to a global variable

row_names = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")[:n]
row_map = dict(zip(row_names, range(n)))

column_names=list(map(str,range(1,n+1)))
column_map=dict(zip(column_names,range(n)))
````

````python
def print_message(message,verbose=True):
    if verbose:
        print(message)
````

````python
def parse_location(location_str):
    if not isinstance(location_str,str):
        print_message("Bad Input. Location must be string.")
        return -1
    
    if len(location_str)!=2:
        print_message("Bad Input. Location must be 2 characters.")
        return -1
    
    row=location_str[0].upper()
    col=location_str[1].upper()
    
    if row not in row_map:
        print_message("Bad Row.")
        return -1

    if col not in column_map:
        print_message("Bad Column.")
        return -1

    return row_map[row],column_map[col]
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
# Write your solution here

def tic_tac_toe_game():

    print("Welcome to a game of tic-tac-toe!")
    print(" ")
    print("Select the dimension of the board")
    
    while True:
        try:
            size = int(input("Enter a number: "))
            break
        except ValueError:
            print("Invalid input. Please enter an integer")

    board_0 = make_board_(size)
    player = player_1
    game_won = False
    
    while not game_won:

        draw_board(board_0)

        print("Player", player, "move:")
        take_move(board_0, player)

        game_won = who_won(board_0)

        if game_won == -1:
            game_won = False

        if player==player_1:
            player=player_2
        else:
            player=player_1


    draw_board(board_0)
    print("The winner is Player:", game_won)
````

````python

def generate_moves(board,player,current_player,depth, keep_all_moves=False):
    if depth==0:
        return list(),score_board(board,player)

    moves=list()
    scores=list()
    
    # Look through the board for pieces belonging to a specific player
    for i in range(size):
        for j in range(size):
            if board[i][j]==current_player:
                
                # For each piece, consider moving left or right
                for move in [left_move,right_move]:
                    
                    # Create a new copy of the board
                    new_board=copy.deepcopy(board)
                    
                    # Check if its a valid move
                    if move_piece(new_board,current_player,(i,j),move,verbose=False):
                        # If so, generate subsequent moves
                        next_moves,score=generate_moves(new_board,player,
                                                        switch_player(current_player),
                                                        depth-1,
                                                        keep_all_moves=keep_all_moves)
                        
                        # Keep track of this move
                        this_move=[(current_player,(i,j),move)]
                        if keep_all_moves:
                            # For our purposes, we usually only need to keep the top move!
                            this_move.extend(next_moves)
                        moves.append(this_move)
                        scores.append(score)

    return moves,scores
````
